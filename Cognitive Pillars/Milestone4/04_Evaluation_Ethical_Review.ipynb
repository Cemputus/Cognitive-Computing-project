## Part 1: Quantitative Performance Evaluation

Evaluate system performance using standard metrics.


```python
# Load results from previous milestones
try:
    df_results = pd.read_csv('../../backend/data/processed/reviews_with_sentiment.csv')
    print(f"✅ Loaded {len(df_results)} results for evaluation")
except FileNotFoundError:
    print("⚠️ Results not found. Please run previous milestones first.")
    df_results = pd.DataFrame()

if not df_results.empty:
    # Calculate basic statistics
    print("\n=== Performance Metrics ===")
    print(f"Total reviews analyzed: {len(df_results)}")
    
    if 'sentiment' in df_results.columns:
        sentiment_dist = df_results['sentiment'].value_counts(normalize=True) * 100
        print(f"\nSentiment Distribution:")
        for sentiment, percentage in sentiment_dist.items():
            print(f"  {sentiment}: {percentage:.2f}%")
        
        # Visualize
        plt.figure(figsize=(10, 6))
        sentiment_dist.plot(kind='bar', color=['green', 'red', 'gray'])
        plt.title('Sentiment Distribution')
        plt.xlabel('Sentiment')
        plt.ylabel('Percentage (%)')
        plt.xticks(rotation=0)
        plt.show()
```

## Part 2: Baseline Comparison

Compare cognitive approach with simple keyword-based baseline.


```python
# Baseline: Simple keyword-based sentiment
def baseline_keyword_sentiment(text):
    """Simple keyword-based sentiment (baseline method)"""
    positive_words = ['good', 'great', 'excellent', 'amazing', 'love', 'best', 'wonderful']
    negative_words = ['bad', 'poor', 'terrible', 'awful', 'hate', 'worst', 'disappointed']
    
    text_lower = str(text).lower()
    pos_count = sum(1 for word in positive_words if word in text_lower)
    neg_count = sum(1 for word in negative_words if word in text_lower)
    
    if pos_count > neg_count:
        return 'positive'
    elif neg_count > pos_count:
        return 'negative'
    else:
        return 'neutral'

# Compare with cognitive approach
if not df_results.empty and 'cleaned_text' in df_results.columns:
    # Apply baseline
    df_results['baseline_sentiment'] = df_results['cleaned_text'].apply(baseline_keyword_sentiment)
    
    # Compare
    if 'sentiment' in df_results.columns:
        comparison = pd.crosstab(df_results['sentiment'], df_results['baseline_sentiment'], 
                                 margins=True, margins_name='Total')
        print("=== Cognitive vs Baseline Comparison ===")
        print(comparison)
        
        # Calculate agreement
        agreement = (df_results['sentiment'] == df_results['baseline_sentiment']).sum()
        agreement_pct = (agreement / len(df_results)) * 100
        print(f"\nAgreement: {agreement}/{len(df_results)} ({agreement_pct:.2f}%)")
        
        print("\n✅ Cognitive approach provides more nuanced analysis with compound scores")
        print("✅ Baseline is limited to keyword matching only")
```

```python
# Analyze potential biases
print("=== Ethical Analysis: Data Bias Assessment ===\n")

if not df_results.empty:
    # Linguistic bias (English vs Luganda)
    print("1. LINGUISTIC BIAS:")
    print("   - Current system optimized for English text")
    print("   - Luganda language support: Limited")
    print("   - Mitigation: Expand preprocessing for Luganda")
    
    # Regional bias
    if 'location' in df_results.columns:
        print("\n2. REGIONAL BIAS:")
        location_dist = df_results['location'].value_counts()
        print(f"   - Data distribution by location:")
        for loc, count in location_dist.items():
            print(f"     {loc}: {count} ({count/len(df_results)*100:.1f}%)")
        print("   - Mitigation: Ensure balanced data collection across regions")
    
    # Source bias
    if 'source' in df_results.columns:
        print("\n3. SOURCE BIAS:")
        source_dist = df_results['source'].value_counts()
        print(f"   - Data distribution by source:")
        for src, count in source_dist.items():
            print(f"     {src}: {count} ({count/len(df_results)*100:.1f}%)")
        print("   - Mitigation: Diversify data sources")
    
    print("\n4. RECOMMENDATIONS:")
    print("   ✅ Expand language support for Luganda")
    print("   ✅ Collect data from diverse regions")
    print("   ✅ Include multiple data sources")
    print("   ✅ Regular bias audits")
    print("   ✅ Fairness metrics monitoring")
```
