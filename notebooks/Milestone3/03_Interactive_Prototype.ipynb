## Part 2: Cognitive Agent Class

Create a cognitive agent that integrates all components for end-to-end processing.


```python
class BusinessIntelligenceAgent:
    """
    Cognitive Agent for Small Business Intelligence Analysis
    Integrates understanding, reasoning, and interaction capabilities
    """
    
    def __init__(self, analyzer, preprocessor, knowledge_graph=None):
        self.analyzer = analyzer
        self.preprocessor = preprocessor
        self.knowledge_graph = knowledge_graph
        
    def process_query(self, user_input):
        """
        Complete cognitive cycle: Understand → Reason → Interact
        
        Args:
            user_input: User's text query or review
            
        Returns:
            Dictionary with analysis results and insights
        """
        # Step 1: Understand - Preprocess and analyze sentiment
        cleaned_text = self.preprocessor.preprocess(user_input)
        sentiment_result = self.analyzer.analyze(cleaned_text)
        
        # Step 2: Reason - Extract insights
        insights = self._extract_insights(cleaned_text, sentiment_result)
        
        # Step 3: Interact - Generate response
        response = self._generate_response(sentiment_result, insights)
        
        return {
            'input': user_input,
            'cleaned_text': cleaned_text,
            'sentiment': sentiment_result,
            'insights': insights,
            'response': response
        }
    
    def _extract_insights(self, text, sentiment_result):
        """Extract key insights from text"""
        insights = {
            'key_phrases': [],
            'entities': [],
            'sentiment_strength': abs(sentiment_result.get('compound', 0))
        }
        
        # Simple keyword extraction (enhance with NER in production)
        business_keywords = ['service', 'delivery', 'product', 'quality', 'price', 'customer']
        for keyword in business_keywords:
            if keyword in text.lower():
                insights['key_phrases'].append(keyword)
        
        return insights
    
    def _generate_response(self, sentiment_result, insights):
        """Generate human-readable response"""
        sentiment = sentiment_result.get('sentiment', 'neutral')
        compound = sentiment_result.get('compound', 0)
        
        if sentiment == 'positive':
            response = f"✅ Positive feedback detected (score: {compound:.2f}). "
        elif sentiment == 'negative':
            response = f"⚠️ Negative feedback detected (score: {compound:.2f}). "
        else:
            response = f"➡️ Neutral feedback (score: {compound:.2f}). "
        
        if insights['key_phrases']:
            response += f"Key topics mentioned: {', '.join(insights['key_phrases'])}. "
        
        response += "This insight can help inform business decisions."
        
        return response

# Initialize agent
agent = BusinessIntelligenceAgent(analyzer, preprocessor, knowledge_graph)
print("✅ Cognitive agent initialized!")
```

```python
# Test queries
test_queries = [
    "Great service, very fast delivery in Nakawa!",
    "Product was damaged on arrival, poor quality",
    "Amazing quality, will definitely buy again",
    "Late delivery but the product itself is good"
]

print("=== Testing Cognitive Agent ===\n")
results = []

for query in test_queries:
    result = agent.process_query(query)
    results.append(result)
    
    print(f"Query: {result['input']}")
    print(f"Sentiment: {result['sentiment']['sentiment']} (score: {result['sentiment'].get('compound', 0):.2f})")
    print(f"Response: {result['response']}")
    print("-" * 60)

# Create results dataframe
df_results = pd.DataFrame(results)
print(f"\n✅ Processed {len(results)} queries successfully")
```
